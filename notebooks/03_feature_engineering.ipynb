```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("data/cars_cleaned.csv")
df.head()
```

```python
df["car_age"] = 2026 - df["year"]
```

```python
df[["year", "car_age"]].head(10)
```

```python
df["mileage_per_year"] = df["mileage(kilometers)"] / df["car_age"]
```

```python
df[["mileage(kilometers)", "car_age", "mileage_per_year"]].head(10)
```

```python
df["engine_volume_liters"] = df["volume(cm3)"] / 1000
```

```python
df[["volume(cm3)", "engine_volume_liters"]].head(10)
```

```python
df["is_newer_car"] = (df["year"] >= 2015).astype(int)
```

```python
df[["year", "is_newer_car"]].head(15)
```

```python
df["is_high_mileage"] = (
    df["mileage(kilometers)"] >= 200000
).astype(int)
```

```python
df[["mileage(kilometers)", "is_high_mileage"]].head(15)
```

```python
df["brand_model"] = (
    df["make"].astype(str) + "_" + df["model"].astype(str)
)
```

```python
df[["make", "model", "brand_model"]].head(10)
```
